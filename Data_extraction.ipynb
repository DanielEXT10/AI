```python
from genericpath import isfile

def get_boards_data(folder):
  files = os.listdir(folder)

  dataframes=[]

  for file in files:
    full_path = os.path.join(folder,file)
    if os.path.isfile(full_path):

      try:

        f = open(full_path)
        data =json.load(f)
        #Run Number
        runs =data['runs'][0]
        run = dict(filter(lambda x:  x[0] == 'runNumber',runs.items() ))

        #Tool Info
        tool = dict(data['equipment'][0])
        fcsn =dict(filter(lambda x: x[0] == 'fcsn',tool.items()))

        #environmental conditions
        environmental =data['results']['environmental']
        env ={key: value.get('value') for key, value in environmental.items()}

        #Temperature exposure
        temp = ['max_hstemp','max_hstemp','hstemp_130','hstemp_150','hstemp_175']
        newDict = dict(filter(lambda x: x[0] in temp,data['results']['general'].items()))
        tempvalues ={k:v['value'] for (k,v) in newDict.items()}


        #Toolset composition
        toolset =data['toolset']
        toolset= [{'fcsn':d['fcsn'],'description': d['description'],'toolsetLevel':d['toolsetLevel'] }for d in toolset if d.get('component') in ['TeleScope PWA','PowerPulse/TeleScope']]
        toolset_dict = {
          'toolset':toolset
            }
        #Final Dictionary
        tooldict = {**fcsn, **tempvalues, **env,**run, **toolset_dict}
        df =pd.DataFrame(tooldict)
        df[['Component','Description', 'level']] =df['toolset'].apply(pd.Series)
        dataframes.append(df)
      except:
        pass
  return dataframes


```

```python
df =get_boards_data('/content/drive/MyDrive/Colab Notebooks/Proyecto de innovacion/Analytics Results/Results')
```
